```python
from pyspark.sql import functions as F

# Bronze já chega com PII anonimizado (cpf_hash, birth_year) no backend
# Nesta camada silver, apenas limpamos e padronizamos

bronze_users = spark.table("workspace.bronze.users")

silver_users = (bronze_users
                .select(
                    "id",
                    "email",
                    "name",
                    "sex",
                    "cpf_hash",
                    "birth_year",
                    "qr_code",
                    "created_at",
                    "updated_at",
                    "_ingested_at"
                )
                .dropDuplicates(["id"])
            )

(silver_users.write
 .format("delta")
 .mode("overwrite")
 .option("overwriteSchema", "true")
 .saveAsTable("silver.users")
 )           

print(f"Silver users: {silver_users.count():,} registros")
```
