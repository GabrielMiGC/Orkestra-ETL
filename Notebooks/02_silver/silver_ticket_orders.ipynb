```python
from pyspark.sql import functions as F
```

```python
bronze_ticket_orders = spark.table("workspace.bronze.ticket_orders")

silver_ticket_orders = (
    bronze_ticket_orders
    .filter("status = 'completed'")
    .withColumn("ordered_date", F.to_date("ordered_at"))
    .withColumn("total_paid", F.col("total_paid").cast("double"))
    .select(
        F.col("id").alias("ticket_order_id"),
        F.col("user_id").alias("user_id"),
        F.col("event_id").alias("event_id"),
        F.col("total_paid").alias("total_paid"),
        F.col("payment_method").alias("payment_method"),
        F.col("status").alias("status"),
        F.col("ordered_at").alias("ordered_date"),
        F.col("created_at").alias("created_at")
    )
)

silver_ticket_orders.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.silver.ticket_orders")
```
