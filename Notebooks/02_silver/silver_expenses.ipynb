```python
from pyspark.sql import functions as F
```

```python
bronze_expenses = spark.table("workspace.bronze.expenses")
bronze_budgets = spark.table("workspace.bronze.budgets")

silver_expenses = (
    bronze_expenses.alias("be")
    .join(bronze_budgets.alias("bb"), F.col("be.budget_id") == F.col("bb.id"))
    .select(
        F.col("be.id").alias("expense_id"),
        F.col("be.budget_id").alias("budget_id"),
        F.col("be.description").alias("description"),
        F.col("be.amount").alias("amount"),
        F.col("be.payment_method").alias("payment_method"),
        F.col("be.paid_at").alias("paid_at"),
        F.col("be.created_at").alias("created_at"),
        F.col("bb.max_amount").alias("max_amount"),
        F.col("bb.currency").alias("currency"),
        )
    )

silver_expenses.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.silver.expenses")
```
