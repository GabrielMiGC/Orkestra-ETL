```python
from pyspark.sql import functions as F
```

```python
bronze_orders_items = spark.table("workspace.bronze.ticket_order_items")
bronze_ticket_orders = spark.table("workspace.bronze.ticket_orders")

silver_ticket_order_items = (
    bronze_orders_items.alias("oi")
    .join(bronze_ticket_orders.alias("to"), F.col("to.id") == F.col("oi.order_id"))
    .select(
        F.col("order_id").alias("order_id"),
        F.col("ticket_id").alias("ticket_id"),
        F.col("unit_price").alias("unit_price")
    )
)

silver_ticket_order_items.write.mode("overwrite").saveAsTable("silver.ticket_order_items")
```
